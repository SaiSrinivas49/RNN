```python
from tensorflow.keras.preprocessing.text  import one_hot

sents = [
    'This is the first sentence',
    'A bunch of idiots',
    'Avengers the age of ultron',
    'The biggest data set is here',
    'Womp womp lil nigga'
]

#Vocabilary size
voc_size = 10000

#One Hot Representation
one_hot_rep = [one_hot(sent,voc_size) for sent in sents]
one_hot_rep
```

```python
#Word Embedding Representaion
from tensorflow.keras.layers import Embedding
from tensorflow.keras.models import Sequential
from tensorflow.keras.utils import pad_sequences
import numpy as np

max_sent_length = 8
embedded_docs = pad_sequences(one_hot_rep,padding='pre',maxlen=max_sent_length)
embedded_docs
```

```python
#Feature representation
dim = 10 #No of features
model = Sequential()
model.add(Embedding(voc_size,dim,input_length=max_sent_length))
model.compile(optimizer='adam',loss='mse')
model.summary()
```

```python
model.predict(embedded_docs)
```
